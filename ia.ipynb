```python
text = "Bonjour ! Je suis EL MOTFI Ouiam, étudiante en deuxième année de master."
```

```python
text = text.lower()
print(text)
```

```python
import string
```

```python
text = text.translate(str.maketrans('', '', string.punctuation))
print(text)
```

Pourquoi est-il important de mettre le texte en minuscules ?

Mettre le texte en minuscules permet de traiter les mots de la même manière et d’éviter de considérer « Bonjour » et « bonjour » comme deux mots différents.

Donnez un exemple de résultat avant et après nettoyage.
Bonjour ! Je suis EL MOTFI Ouiam,
bonjour je suis el motfi ouiam
    

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)
print(tokens)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('french'))

print(stop_words)
```

```python
tokens_filtrés = [word for word in tokens if word not in stop_words]

print(tokens_filtrés)
```

```python
stop_words.add('bonjour')
```

```python
tokens_filtrés = [word for word in tokens if word not in stop_words]

print(tokens_filtrés)
```

Que représentent les stopwords ?

Les stopwords sont des mots fréquents qui apportent généralement peu d’information pour l’analyse d’un texte, comme « je », « de », « en » ou « les ».

Quelle proportion du texte original est supprimée après leur retrait ?

Le texte original contient 12 mots. Après la suppression des stopwords, il reste 8 mots.

Nombre de mots supprimés : 4

Proportion supprimée :

(4 / 12) × 100 = 33,33 %

Donc, 33,33 % des mots du texte original ont été supprimés.

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
tokens_lemmatisés = [lemmatizer.lemmatize(word) for word in tokens_filtrés]

print(tokens_lemmatisés)
```

```python
print("Avant lemmatisation :", tokens_filtrés)
print("Après lemmatisation :", tokens_lemmatisés)
```

Quelle est la différence entre stemming et lemmatisation ?

Le stemming consiste à réduire un mot à sa racine en supprimant certains suffixes..

La lemmatisation consiste à réduire un mot à sa forme de base correcte .

Pourquoi cette étape est-elle importante avant l’analyse sémantique ?

La lemmatisation permet de regrouper les différentes formes d’un même mot. Cela réduit le nombre de mots différents et facilite l’analyse du sens du texte.
